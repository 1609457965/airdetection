```python
# 读取2.csv
import pandas as pd
data=pd.read_excel('2.xls', header=None)

rows = []
# Skip the header row (row 0) 
for idx, row in data.iloc[1:].iterrows():
    date = row[0]
    weather = row[1]
    temp = row[2]
    wind = row[3]
    weather_day, weather_night = weather.split('/')
    temp_day, temp_night = temp.split('/')
    wind_day, wind_night = wind.split('/')

    rows.append([date, '白天', weather_day.strip(), temp_day.strip(), wind_day.strip()])
    rows.append([date, '夜间', weather_night.strip(), temp_night.strip(), wind_night.strip()])

# 构建DataFrame
df = pd.DataFrame(rows, columns=['日期', '时间段', '天气状况', '气温', '风力风向'])
df
```

```python
#气温删掉℃符号 , 白天是0 ,夜晚是1
df['气温'] = df['气温'].str.replace('℃', '').astype(float)
df['时间段'] = df['时间段'].map({'白天': 0, '夜间': 1})
df
```

```python
# 保存为新的Excel文件
df.to_csv('2分.csv', index=False, header=False)
```
