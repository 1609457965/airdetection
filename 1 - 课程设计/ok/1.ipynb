```python
#读取 全.csv 文件
import pandas as pd
import numpy as np
#读取 全.csv 文件
df = pd.read_csv('全.csv', encoding='utf-8')
#读取缺.csv 文件
df2 = pd.read_csv('缺.csv', encoding='utf-8')
#将所有0值替换为以"点位名称"列分组,"SO2"	"NO2"	"PM10"	"CO"	"O31小时"	"O38小时"	"PM2.5"列数据的非0最小值, 以及所有数据四舍五入为整数

cols = ['SO2', 'NO2', 'PM10', 'CO', 'O31小时', 'O38小时', 'PM2.5']

for col in cols:
    # 按点位名称分组，取该组下非0最小值
    group_min = df.groupby('点位名称')[col].transform(lambda x: x[x != 0].min())
    # 替换 0 值为对应分组的非0最小值
    df[col] = df[col].replace(0, np.nan).fillna(group_min)
    # 四舍五入并转为整数
    df[col] = df[col].round().astype(int)
df.to_csv('处理后的全.csv', index=False, encoding='utf-8')
```

```python
# 将"污染等级"列的值替换为数字
pollution_mapping = {
    '优': 0,
    '良': 1,
    '轻微污染': 2, '轻微': 2,
    '轻度污染': 3, '轻度': 3,
    '中度污染': 4, '中度': 4,
    '中度重污染': 5,
    '重度污染': 6, '重度': 6,
    '严重污染': 7, '重污染': 7,'严重': 7
}
# 分别读取"缺.csv" 和"处理小数和0后的全.csv"文件
df2 = pd.read_csv('缺.csv', encoding='utf-8')
df = pd.read_csv('处理小数和0后的全.csv', encoding='utf-8')
# 将"首要污染物"列的值替换为数字
df['首要污染物'] = df['首要污染物'].replace(pollution_mapping)
df.to_csv('编码后全.csv', index=False, encoding='utf-8')
# 将"首要污染物"列的值替换为数字
df2['首要污染物'] = df2['首要污染物'].replace(pollution_mapping)
df2.to_csv('编码后缺.csv', index=False, encoding='utf-8')  

```

```python
# 读取并将两表合并
import pandas as pd
df = pd.read_csv('编码后全.csv', encoding='utf-8')
df2 = pd.read_csv('编码后缺.csv', encoding='utf-8')
df_combined = pd.concat([df, df2], ignore_index=True)
# 将合并后的数据保存到新的CSV文件
df_combined.to_csv('1处理的数据.csv', index=False, encoding='utf-8')
```
